```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("===Zomato Sales Data Analysis===")
print("sales data shape : ", sales_data.shape)
print("sales data for first three row" , sales_data[: ,1:])
print("sales data for first three row" , sales_data[:3 ,1:])
```

```python
# total sales per year 

print(np.sum(sales_data , axis =0))  # column wise sum
yearly_total = np.sum(sales_data[: , 1:] , axis =0)
print("yearly total sales : " , yearly_total)

# Minimum sales per restaurant

min_sales =np.min(sales_data[: , 1: ] , axis=1)
print("minimum sales per restaurant : " , min_sales)


#maximum sales per year

max_sales = np.max(sales_data[: , 1:] , axis = 0 )
print(max_sales)

# average sales per restaurant

avg_sales = np.mean(sales_data[: , 1:] , axis =1)
print("average sales per restaurant : " , avg_sales)

#cumsumative sales per year
cumsum_sales = np.cumsum(sales_data[: , 1:] , axis =1)
print("cumulative sales per restaurant : " , cumsum_sales)


# graphical representation

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum_sales , axis =0))
plt.title("average cumulative sales across all restaurants")
plt.xlabel("Years ")
plt.ylabel("Average Cumulative Sales")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,6])

print("Vector Addition :" , vector1 + vector2)

print(" \n Vector multiplication  :" , vector1 * vector2)

print(" \n Vector dot product  :" , np.dot(vector1 , vector2))

angle = np.arccos(np.dot(vector1 , vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print( "The angle between the v1 & v2 is :",angle)
```
